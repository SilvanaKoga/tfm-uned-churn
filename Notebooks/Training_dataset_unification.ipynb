```python
import pandas as pd

import seaborn as sb
import matplotlib.pyplot as plt
```

```python
DATA_PATH = "../Data/Training_dataset"
```

```python
NINETY_DAYS_T1_PATH = "churn_dataset_90_days1.csv"
NINETY_DAYS_T2_PATH = "churn_dataset_90_days2.csv"
NINETY_DAYS_T3_PATH = "churn_dataset_90_days3.csv"
ONE_HUNDRED_T1_PATH = "churn_dataset_100_days1.csv"
ONE_HUNDRED_T2_PATH = "churn_dataset_100_days2.csv"
ONE_HUNDRED_ONE_T1_PATH = "churn_dataset_110_days1.csv"
ONE_HUNDRED_ONE_T2_PATH = "churn_dataset_110_days2.csv"
```

```python
churn_dataset_90_days1 = pd.read_csv(DATA_PATH + "/" + NINETY_DAYS_T1_PATH) 
churn_dataset_90_days2 = pd.read_csv(DATA_PATH + "/" + NINETY_DAYS_T2_PATH)
churn_dataset_90_days3 = pd.read_csv(DATA_PATH + "/" + NINETY_DAYS_T3_PATH)
churn_dataset_100_days1 = pd.read_csv(DATA_PATH + "/" + ONE_HUNDRED_T1_PATH)
churn_dataset_100_days2 = pd.read_csv(DATA_PATH + "/" + ONE_HUNDRED_T2_PATH)
churn_dataset_110_days1 = pd.read_csv(DATA_PATH + "/" + ONE_HUNDRED_ONE_T1_PATH)
churn_dataset_110_days2 = pd.read_csv(DATA_PATH + "/" + ONE_HUNDRED_ONE_T2_PATH)
```
