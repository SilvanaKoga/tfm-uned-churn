```python
import pandas as pd
import numpy as np
```

```python
DATA_PATH = "../Data/olist_post_eda.csv"

olist_df = pd.read_csv(DATA_PATH)
```

```python
olist_df["freight_ratio"].isna().sum()
```

```python
# Drop rows where approved at, price, payment_type and productid is missing. They are part of the feature.

olist_df = olist_df.dropna(subset= ["order_approved_at", "price", "product_id", "payment_type"])

# fill the missing values of review score with 0. 0 means they didn't comment.

olist_df["review_score"] = olist_df["review_score"].fillna(0)

# Drop the review_comment_title since more than 88% of its rows are missing and do not provide information. The same for review_comment_message that has 58% of missing value
# review_id and review_creation_day will not be used
drop_columns = ["review_comment_title", "review_comment_message", "review_id", "review_creation_date", "review_answer_timestamp", "product_category_name", "product_name_lenght", "product_description_lenght", "product_photos_qty", "product_weight_g", "product_length_cm", "product_height_cm", "product_width_cm"]

olist_df = olist_df.drop(columns= [col for col in drop_columns if col in olist_df.columns], errors= "ignore")

# Clean the delivery delays column "dates_diff"

olist_df["dates_diff"] = olist_df["dates_diff"].fillna(0)

olist_df["price"] = olist_df["price"].fillna(0)

olist_df["freight_ratio"] = olist_df["freight_ratio"].fillna(0)

olist_df["product_category_name_english"] = olist_df["product_category_name_english"].fillna("unknown")

olist_df["product_cat"] = olist_df["product_cat"].fillna("unknown")
```
